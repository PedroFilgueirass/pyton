```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("atividade_5.csv", sep=";", decimal=",")
```

```python
df.head()
```

```python
df = df.dropna(axis=1, how="all")
```

```python
print(df.columns)
```

```python
df["2024"] = pd.to_numeric(df["2024"], errors="coerce")
```

```python
df_ordenado = df.sort_values(
    by="2024",
    ascending=False
)
```

```python
print(df_ordenado[["Sigla", "2024"]])
```

```python
df["1991"] = pd.to_numeric(
    df["1991"],
    errors="coerce"
)
```

```python
df["Melhora"] = df["2024"] - df["1991"]
```

```python
maior_melhora = df.loc[
    df["Melhora"].idxmax()
]
```

```python
print(
    maior_melhora[
        ["Sigla", "1991", "2024", "Melhora"]
    ]
)
```

```python
estados_pioraram = df[
    df["Melhora"] < 0
]
```

```python
print(
    estados_pioraram[
        ["Sigla", "1991", "2024", "Melhora"]
    ]
)
```

```python
if estados_pioraram.empty:
    print("Nenhum estado teve piora no IDH.")
else:
    print("Existem estados que tiveram piora.")
```

```python
anos = [
    c for c in df.columns
    if str(c).isdigit()
]

print(anos)
```

```python
id_vars = [
    c for c in df.columns
    if c not in anos
]
```

```python
df_longo = df.melt(
    id_vars=id_vars,
    value_vars=anos,
    var_name="Ano",
    value_name="IDH"
)
```

```python
df_longo["Ano"] = df_longo["Ano"].astype(int)

df_longo["IDH"] = pd.to_numeric(
    df_longo["IDH"],
    errors="coerce"
)
```

```python
df_longo.head()
```

```python
mg = df_longo[
    df_longo["Sigla"] == "MG"
]
```

```python
mg = mg.sort_values("Ano")
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(
    mg["Ano"],
    mg["IDH"],
    marker="o",
    linewidth=2
)

ax.set_title(
    "Evolução do IDH de Minas Gerais"
)

ax.set_xlabel("Ano")
ax.set_ylabel("IDH")

ax.set_ylim(0.3, 0.9)

ax.grid(True)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 7))
```

```python
for sigla, grupo in df_longo.groupby("Sigla"):

    grupo = grupo.sort_values("Ano")

    ax.plot(
        grupo["Ano"],
        grupo["IDH"],
        marker="o",
        markersize=3,
        linewidth=1.5,
        label=sigla
    )
```

```python
ax.set_title(
    "Evolução do IDH por estado (1991–2024)"
)

ax.set_xlabel("Ano")
ax.set_ylabel("IDH")

ax.set_ylim(0.3, 0.9)
```

```python
ax.legend(
    ncol=3,
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    title="UF"
)
```

```python
fig.tight_layout()

plt.show()
```
